```python
from illustrated_agents.chapters.ch2 import LLM
```

```python
llm = LLM(model="gemma3:12B", temperature=1)
```

```python
# Example of one-shot prompting
prompt_example = "I saw 6 flamingos, 2 flew away, 1 hid behind the tree, how many can I see? Be concise"
prompt_answer = "3"

# The query we want to ask the model
query = "I saw 9 penguins, 2 slid in the water and disappeared from sight while 4 waddled up from the shore. How many I can see?"
```

```python
# Messages
messages = [
    {"role": "user", "content": prompt_example},
    {"role": "assistant", "content": prompt_answer},
    {"role": "user", "content": query}
]
```

```python
# Generate response - Wrong answer
response = llm.generate(messages)
print(response)
```

```python
# Correct Answer
prompt_answer = "You saw 6 flamigos 2 flew away. 6 - 2 = 4. Then, 1 flamingo hid. You see now 4 - 1 = 3."
# Messages
messages = [
    {"role": "user", "content": prompt_example},
    {"role": "assistant", "content": prompt_answer},
    {"role": "user", "content": query}
]
response = llm.generate(messages)
print(response)
```

```python
# Zero-shot prompting
messages = [
    {"role": "user", "content": query + " Let's think step by step."},
]
response = llm.generate(messages)
print(response)
```
